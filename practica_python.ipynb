## Tarea 2 - Listas, bucles y condicionales

Recorremos una lista de precios y clasificamos cada precio según su valor

```python
precios = [100 , 150 , 200.5 , 50 , 90]

for p in precios:
  if p > 100:
    print(f"{p} es caro")
  else:
    print (f"{p} es barato")
```

## Tarea 4 - Funciones

Creamos una función que calcula la suma, cantidad y promedio de una lista de números

```python
def resumen_estadistico(numeros):
  total = sum(numeros)
  cantidad = len(numeros)
  promedio = total/cantidad

  return {"suma": total ,
          "cantidad":cantidad,
          "promedio":promedio

          }

resultado = resumen_estadistico([100,250,350,400,190])

print(resultado)
```
